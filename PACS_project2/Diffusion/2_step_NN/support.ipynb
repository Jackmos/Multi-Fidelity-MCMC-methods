```python
r2_HF_df = pandas.read_csv('./Output_new/r2_HF_lhs_transformation.txt',  sep='\t')
mse_HF_df = pandas.read_csv('./Output_new/mse_HF_lhs_transformation.txt', sep = '\t')
print(r2_HF_df.index)
```

```python
#########################     DATA     ##########################
r2_HF_df = pandas.read_csv('./Output_new/r2_HF_lhs_transformation.txt',  sep='\t')
mse_HF_df = pandas.read_csv('./Output_new/mse_HF_lhs_transformation.txt', sep = '\t')
HF_data = np.loadtxt("../DATA_shear_cube/Data_Train_bases_young/HF_num_data.txt").size
Nlf_models = np.loadtxt("../DATA_shear_cube/Data_Train_bases_young/N_bases.txt").astype(int)
#r2_HF_df.index = [str(num) for num in [HF_data]]
r2_HF_df.index=range(HF_data)
mse_HF_df.index = range(HF_data)
Nlf_models_str = [str(num) for num in Nlf_models]
HF_data_strr = [str(num) for num in range(2)]
style_lines = [':*','--^','-.o','-s',':D','--x','-*',':v','-o']

with open('./Output_new/U_HF_list_transformation.data', 'rb') as filehandle:
    U_HF_list = pickle.load(filehandle)



r2_HF_df.round(5)
mse_HF_df.round(5)
```

```python
#########################     PLOTS     ##########################
lines = mse_HF_df.plot.line(y = Nlf_models_str, style = style_lines, markersize = 4)
plt.title('MSE')
plt.legend(fontsize = 8.5, loc ='upper right')


lines = r2_HF_df.plot.line(y = Nlf_models_str, style = style_lines, markersize = 4)
plt.title('R^2')
plt.legend(fontsize = 8.5, loc ='lower right')


lines = mse_HF_df.T[:10].plot.line(style = style_lines, markersize = 4)
plt.title('MSE')
plt.legend(fontsize = 8.5, loc ='upper right')


lines = r2_HF_df.T[:10].plot.line(style = style_lines,markersize = 4)
plt.title('R^2')
plt.legend(fontsize = 8.5, loc ='lower right')


#########################     ZOOM IN of PLOTS     ##########################
lines = mse_HF_df.plot.line(y = Nlf_models_str, style = style_lines, markersize = 4)
plt.title('MSE')
plt.legend(fontsize = 7, loc ='upper right')
plt.ylim(0,0.01)
plt.show()

lines = r2_HF_df.plot.line(y = Nlf_models_str, style = style_lines, markersize = 4)
plt.title('R^2')
plt.legend(fontsize = 7.5, loc ='lower right')
plt.ylim(0.7,1.0)
plt.show()

lines = mse_HF_df.T[:10].plot.line(style = style_lines, markersize = 4)
plt.title('MSE')
plt.legend(fontsize = 6.5, loc ='lower left')
plt.ylim(0,0.01)
plt.show()

lines = r2_HF_df.T[:10].plot.line(style = style_lines,markersize = 4)
plt.title('R^2')
plt.legend(fontsize = 7.5, loc ='upper left')
plt.ylim(0.7,1.0)
plt.show()
```
